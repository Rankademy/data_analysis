```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import os

# 1. 데이터 불러오기
df = pd.read_csv("data/player_full_data.csv")
print(f"✅ 원본 데이터 shape: {df.shape}")
```

```python
# 2. 수치형 피처만 선택
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# 3. 결측치 처리 (간단하게 0으로 대체)
numeric_df = numeric_df.fillna(0)

# 4. 스케일링
scaler = StandardScaler()
scaled_array = scaler.fit_transform(numeric_df)
scaled_df = pd.DataFrame(scaled_array, columns=numeric_df.columns)
```

```python
# 5. 저장
os.makedirs("data", exist_ok=True)
scaled_df.to_csv("data/player_preprocessed.csv", index=False, encoding='utf-8-sig')

print("📁 전처리 완료 및 저장: data/player_preprocessed.csv")
scaled_df.head()
```
